```python
import geopandas as gpd
import pandas as pd
import folium
import datetime
```

```python
fmap = folium.Map(location=[-23.5, -46.6], zoom_start=10)

GDF_ciclorrotas = gpd.read_file('C:\\Users\\jojos\\projetos_estagio\\source\\bike_science\\files\\rede_cicloviaria_historico\\Ciclorrotas.shp')
GDF_ciclovias = gpd.read_file('C:\\Users\\jojos\\projetos_estagio\\source\\bike_science\\files\\rede_cicloviaria_historico\\Ciclovias.shp')
GDF_ciclo_historico = pd.concat([GDF_ciclorrotas, GDF_ciclovias])
GDF_ciclo_historico = GDF_ciclo_historico.reset_index()

GDF_ciclo_historico['inauguracao'] = pd.to_datetime(GDF_ciclo_historico['inauguracao'])

ciclo_08_12 = GDF_ciclo_historico[GDF_ciclo_historico['inauguracao'] < datetime.datetime(2013,1,1)]

start = datetime.datetime(2013,1,1)
end = datetime.datetime(2017,1,1)
ciclo_13_16 = GDF_ciclo_historico[(GDF_ciclo_historico['inauguracao']>=start) & (GDF_ciclo_historico['inauguracao']<end)]

start = datetime.datetime(2017,1,1)
end = datetime.datetime.now()
ciclo_17_24 = GDF_ciclo_historico[(GDF_ciclo_historico['inauguracao']>=start) & (GDF_ciclo_historico['inauguracao']<end)]

ciclo_08_12 = ciclo_08_12.set_crs('epsg:4326')
ciclo_13_16 = ciclo_13_16.set_crs('epsg:4326')
ciclo_17_24 = ciclo_17_24.set_crs('epsg:4326')
```

```python
folium.GeoJson(
    name='Ciclovias de 2008-2012',
    data=ciclo_08_12['geometry'],
    style_function = lambda x: { "color": ("#0000ff") }
).add_to(fmap)

folium.GeoJson(
    name='Ciclovias de 2013-2016',
    data=ciclo_13_16['geometry'],
    style_function = lambda x: { "color": ("#00ff00") }
).add_to(fmap)

folium.GeoJson(
    name='Ciclovias de 2017-2024',
    data=ciclo_17_24['geometry'],
    style_function = lambda x: { "color": ("#ff0000") }
).add_to(fmap)

folium.LayerControl().add_to(fmap)
```
