```python
import requests
from shapely.geometry import LineString



def graphopper_request(url):
    p_latitude = row['polygons_center'].y
    p_longitude = row['polygons_center'].x
    c_latitude = row['circle_center'].y
    c_longitude = row['circle_center'].x
    response = requests.get(url.format(p_latitude, p_longitude, c_latitude, c_longitude))
    if response.status_code == 200:
        json_data = response.json()
        return json_data
    else:
        raise Exception(f"Error: {response.status_code}")

def convert_list_columns_to_str(df):
    # Convert any list-type columns to strings
    for col in df.columns:
        if df[col].apply(type).eq(list).any():
            df[col] = df[col].astype(str)
    return df

def save_json(gdf, json_filename):
    # Save the GeoDataFrame to a GeoJSON file
    gdf.to_file(json_filename, driver='GeoJSON')
```
