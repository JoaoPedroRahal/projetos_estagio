```python
import folium
import pandas as pd
import geopandas as gpd
import json
from unidecode import unidecode
```

```python
# Create the world map
mp = folium.Map(location=[-15, -50], zoom_start=3.5 )

mp
```

```python
# Extract the cities' coordinates and names 
with open('br.json', encoding = 'utf-8') as file:
    city_data = json.load(file)

city_df = pd.DataFrame(city_data)

# Remove special characters from the names
city_df['city'] = city_df['city'].apply(lambda x: unidecode(x))

#display(city_df)
```

```python
marker_layer = folium.map.FeatureGroup(name='Markers')

for city_info in city_data:
    city = city_info['city']
    lat = float(city_info['lat'])
    lon = float(city_info['lng'])
    
    if city_info['capital'] == 'admin' or city_info['capital'] == 'primary':
        marker = folium.Marker(location=[lat, lon], popup=city)
        marker.add_to(marker_layer)
    
marker_layer.add_to(mp)
    
mp
```

```python
# Import the Dataframe containing the state's polygonal data and the coffee production
geojson_path = 'coffee_brazil.geojson'
coffee_brazil_df = gpd.read_file(geojson_path)
```

```python
# Create the bin suiting the choropleth needs
bins = list(coffee_brazil_df["Coffee_Production"].quantile([0, 0.125, 0.25, 0.5, 0.75, 1]))
```

```python
# Overlay the colored polygons with Brazil's map
folium.Choropleth(
    geo_data=geojson_path,
    name='Choropleth',
    data=coffee_brazil_df,
    columns=['State', 'Coffee_Production'],
    key_on='feature.properties.State',
    fill_color='YlGn',
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name='Coffee Production in millions of bags (60kg) in the year 2021',
    bins = bins
).add_to(mp)

mp
```

```python
# DataFrame for connecting only capitals
capital_df = city_df[city_df['capital'] == 'admin']
capital_df = capital_df.reset_index()

# Create a FeatureGroup to hold the polylines
line_layer = folium.FeatureGroup(name='Polylines')

# Iterate over the cities and connect the coordinates with polylines
for i in range(len(capital_df)):
    city_A = capital_df.loc[i, 'city']
    city_B = capital_df.loc[(i + 1) % len(capital_df), 'city']
    coor_A = (float(capital_df.loc[i, 'lat']), float(capital_df.loc[i, 'lng']))
    coor_B = (float(capital_df.loc[(i + 1) % len(capital_df), 'lat']), float(capital_df.loc[(i + 1) % len(capital_df), 'lng']))
    line = folium.PolyLine([coor_A, coor_B], color='blue')
    line.add_to(line_layer)

# Add the FeatureGroup to the map
line_layer.add_to(mp)

mp
```

```python
folium.LayerControl().add_to(mp)
```

```python
# Save the map
mp.save('brazil_layers.html')
```
